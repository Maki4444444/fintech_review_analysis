# Task 1: Data Collection

**Objective:** Scrape Google Play Store reviews for three Ethiopian banks and save the raw dataset for preprocessing.

**Banks:**
- Commercial Bank of Ethiopia (CBE) — `com.combanketh.mobilebanking`
- Bank of Abyssinia (BOA) — `com.boa.boaMobileBanking`
- Dashen Bank — `com.dashen.dashensuperapp`

**Target:** 400+ reviews per bank (1,200 total)  
**Output:** `data/raw/reviews_raw.csv`

```python
import sys
import os
sys.path.append(os.path.abspath(".."))

import pandas as pd
from scripts.scrape import scrape_reviews
```

## 1. Scraping

We use `google-play-scraper` to pull reviews for each bank. `count=600` gives us buffer above the 400-review minimum after any losses from deduplication and null removal downstream. `lang='en'` and `country='et'` target the Ethiopian English-language Play Store.

```python
APP_IDS = {
    "Commercial Bank of Ethiopia": "com.combanketh.mobilebanking",
    "Bank of Abyssinia": "com.boa.boaMobileBanking",
    "Dashen Bank": "com.dashen.dashensuperapp",
}

raw_df = scrape_reviews(APP_IDS, count=600)
print(f"Total raw reviews scraped: {len(raw_df)}")
raw_df.head()
```

## 2. Raw Data Snapshot

Quick inspection before saving — shape, dtypes, nulls, and per-bank counts.

```python
print(f"Shape: {raw_df.shape}")

print("\n=== Reviews per bank ===")
print(raw_df["bank"].value_counts())

print("\n=== Null values per column ===")
print(raw_df.isnull().sum())

print("\n=== Duplicate review IDs ===")
print(f"Total duplicates: {raw_df.duplicated(subset='id').sum()}")

print("\n=== Date range ===")
print(f"Earliest: {raw_df['date'].min()}")
print(f"Latest:   {raw_df['date'].max()}")
```

## 3. Save Raw Data

Saved to `data/raw/reviews_raw.csv`. This is the unprocessed output — cleaning, deduplication, and normalization happen in `02_preprocessing.ipynb`. This file is listed in `.gitignore` and will not be committed to GitHub.

```python
os.makedirs("../data/raw", exist_ok=True)
raw_df.to_csv("../data/raw/reviews_raw.csv", index=False)
print(f"Saved {len(raw_df)} raw reviews to data/raw/reviews_raw.csv")
```
